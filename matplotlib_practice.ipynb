The intent of this notebook is to work through an exercise for matplot lib. This will go through basic plotting functionality 

```python
# import the relevant libraries 
import matplotlib.pyplot as plt
%matplotlib inline

from matplotlib.ticker import MaxNLocator
```

```python
# create the data that we will be plotting 
apl_price = [93.95, 112.15, 104.05, 144.85, 169.49]
ms_price = [39.01, 50.29, 57.05, 69.98, 94.39]
year = [2014, 2015, 2016, 2017, 2018]
```

```python
# create a basic line plot 
plt.plot(year, apl_price, 'k',
         year, ms_price)

plt.xlabel('Year')
plt.ylabel('Stock Price')

# below shows how to define min and max for axis
# plt.axis([xmin, xmax, ymin, ymax])
# plt.axis([2013, 2019, 20, 180])

plt.show()
```

```python
# now we will create a figure with a grid and customize the axis locators

# 6.4 & 4.8 are defualt for width and height 
fig_1 = plt.figure(1,figsize=(20,4.8))

# the numbers define how many charts you want to display and where the chart is positioned in the structure 
# we created a 1.2 grid, one row, two columns
apple_price = fig_1.add_subplot(121)
microsoft_price= fig_1.add_subplot(122)

apple_price.plot(year, apl_price)
#adjust the xaxis
apple_price.xaxis.set_major_locator(MaxNLocator(integer=True))


microsoft_price.scatter(year, ms_price)
microsoft_price.xaxis.set_major_locator(MaxNLocator(integer=True))

plt.show()
```

```python
# now we will show how to work with different grid sizes
# and populate different subplots within the figure

# parameter in subplots function is number of rows, second is number of columns
fig_2, axes = plt.subplots(2,2,figsize=(20,5))

# refer to a sub plot to add something to it, this refers to first subplot
axes[0,1].scatter(year, ms_price)
# if we increase rows to be 2 x 2, we need to reference row index, and plot index 
axes[1, 0].plot(year, apl_price)

plt.show()
```
